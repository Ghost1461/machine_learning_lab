# Filtering

```python
import pandas as pd 
dataframe = pd.read_csv("titanic.csv")
filtered = dataframe.loc[#koşula uyan satırları seçer
    (dataframe["Sex"] == "female") & #dataframe.loc[:, "Sex"] == "female" -> aynı şeyi yapar ama olayı uzatır
    (dataframe["Age"] >= 63)
]
print(filtered)
```

# Replace

```python
import pandas as pd 
replaced_dataframe = pd.read_csv("titanic.csv")
replaced_dataframe["Sex"] = replaced_dataframe["Sex"].replace("female","woman")
#Sex sütununu seçip female'i woman olarak değiştirip, sonrasında asıl dataframedeki Sex sütununa eşitliyoruz.
```

# GroupBy

```python
import pandas as pd 
dataframe = pd.read_csv("titanic.csv")
print(dataframe["Age"].min())
print(dataframe["Age"].max())
print(dataframe["Age"].mean())
print(dataframe["Age"].sum())
print(dataframe["Age"].count())


print(dataframe.groupby("Sex")["Age"].mean())
print(dataframe.groupby("Sex").mean(numeric_only=True))
```

# Drop Duplicates

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Alex", "John", "Alex", "Emma", "John", "Taha", "Hamsi"],
    "Age": [20, 25, 20, 30, 25, 23, 61],
    "City": ["Izmir", "Ankara", "Izmir", "Istanbul", "Ankara","Trabzon","Trabzon"]
})
print(df.duplicated())#daha önce görülen satır nerede tekrar göründü
print((df.duplicated().sum()))
df.drop_duplicates()

#df.drop_duplicates(subset=["Name"]) Name'i tekrar edenleri buluyor

'''
sadece son kaydı tutuyor Name'i tekrar edenlerden
df.drop_duplicates(
    subset=["Name"],
    keep="last"
)
'''
```

# Concatenate

```python
data_a = {
'id': ['1', '2', '3'],
'first': ['Alex', 'Amy', 'Allen'],
'last': ['Anderson', 'Ackerman', 'Ali']}
dataframe_a = pd.DataFrame(data_a, columns = ['id', 'first', 'last'])

data_b = {
'id': ['4', '5', '6'],
'first': ['Billy', 'Brian', 'Bran'],
'last': ['Bonder', 'Black', 'Balwner']}

dataframe_b = pd.DataFrame(data_b, columns = ['id', 'first', 'last'])

row_dataframe = pd.concat(
    [dataframe_a,dataframe_b],
    axis=0,
    ignore_index=True #index'in kaldığı yerden devam etmesi için
)

column_dataframe = pd.concat( #aynı sütun varsa onları da ekler peşine
    [dataframe_a,dataframe_b],
    axis = 1
)
```
