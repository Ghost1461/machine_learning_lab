# Range Normalization


```python
import numpy as np
from sklearn import preprocessing

features = np.array([[-500.5],
[-100.1],
[0],
[100.1],
[900.9]])

minmax_scale = preprocessing.MinMaxScaler(feature_range=(0,1))

scaled_feature = minmax_scale.fit_transform(features)

print(scaled_feature)
```

# Standart Deviation Normalization

```python
import numpy as np
from sklearn import preprocessing

features = np.array([[-1000.1],
[-200.2],
[500.5],
[600.6],
[9000.9]])

mean = np.mean(features)
std = np.std(features)

sd_scale = preprocessing.StandardScaler()

scaled_feature = sd_scale.fit_transform(features)

mean_scaled = np.mean(scaled_feature)
std_scaled = np.std(scaled_feature)

print(scaled_feature)
print(mean)
print(std)
print(mean_scaled)
print(std_scaled)
```

# Detecting Outliers (IQR Method)

```python
import numpy as np

features = np.array([
    [10],
    [12],
    [11],
    [13],
    [15],
    [14],
    [1000],
    [-100]
])

Q1 = np.percentile(features, 25)
Q3 = np.percentile(features, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = features[(features < lower_bound) | (features > upper_bound)]

print(outliers)
```

# Fill in missing values(Mean Imputation)

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer

features = np.array([[1.1, 11.1],
[2.2, 22.2],
[3.3, 33.3],
[4.4, 44.4],
[np.nan, 55]])

imputer = SimpleImputer(strategy="mean")
'''
"mean"	Ortalama ile doldurur
"median"	Medyan ile doldurur
"most_frequent"	En sık geçen değerle doldurur
"constant"	Belirlediğimiz sabit değerle doldurur
'''
features_imputed = imputer.fit_transform(features)

print(features_imputed)
```

# Predicting Missing Categorical Values With KNN

```python
import numpy as np
import pandas as pd

from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Age": [20, 25, 45, 50, 22, 48],
    "Income": [2000, 3000, 7000, 8000, 2500, 7500],
    "Category": [
        "Budget", "Budget",
        "Premium", "Premium",
        np.nan, np.nan
    ]
})

known = df[df["Category"].notna()]#Eksik olmayan değerleri seçer.
missing = df[df["Category"].isna()]#Eksik değerleri seçer.

X_train = known[["Age", "Income"]]
y_train = known["Category"]

X_missing = missing[["Age", "Income"]]

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_missing_scaled = scaler.transform(X_missing)#yeni gözlemlerin de eğitim verisinden öğrendiğimiz aynı ortalama ve standart sapmayla dönüştürülmesi gerekiyor.

knn = KNeighborsClassifier(n_neighbors=3)

knn.fit(X_train_scaled, y_train)

predictions = knn.predict(X_missing_scaled)

df.loc[df["Category"].isna(), "Category"] = predictions

print(df)
```

# Undersampling, Oversampling and SMOTE

```python
import numpy as np
from sklearn.datasets import load_iris
from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import RandomOverSampler

iris = load_iris()

X = iris.data
y = iris.target

print(np.unique(y, return_counts=True))#hangi sınıftan kaç tane var?

# Setosa sınıfından ilk 10 örneği al
setosa_indices = np.where(y == 0)[0][:10]

# Diğer iki sınıfın bütün örneklerini al
other_indices = np.where(y != 0)[0]

# İndeksleri birleştir
indices = np.concatenate([setosa_indices, other_indices])

X_imbalanced = X[indices]

# Setosa = 0, Not Setosa = 1
y_imbalanced = (y[indices] != 0).astype(int)

print(np.unique(y_imbalanced, return_counts=True))

#Undersampling
under_sampler = RandomUnderSampler(random_state=42)

X_under, y_under = under_sampler.fit_resample(
    X_imbalanced,
    y_imbalanced
)

print(np.unique(y_under, return_counts=True))

#Oversampling(Potansiyel overfitting riski)
over_sampler = RandomOverSampler(random_state=42)

X_over, y_over = over_sampler.fit_resample(
    X_imbalanced,
    y_imbalanced
)

print(np.unique(y_over, return_counts=True))

#SMOTE(en yakın n komşudan sentetik örnekler oluşturur)
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42, k_neighbors=3)

X_smote, y_smote = smote.fit_resample(
    X_imbalanced,
    y_imbalanced
)

print(np.unique(y_smote, return_counts=True))
```
